### Listing Price Distribution

```python
price_bins = [0, 200, 500, 800, float('inf')]

price_labels = [
    'Under 200',
    '200–499',
    '500–799',
    '800 and above'
]

price_groups = pd.cut(
    df['price'],
    bins=price_bins,
    labels=price_labels,
    right=False
)

price_counts = price_groups.value_counts().sort_index()

price_counts.plot(
    kind='bar',
    figsize=(9, 5),
    title='Airbnb Listings by Price Category',
    xlabel='Price Category',
    ylabel='Number of Listings',
    rot=15
)

plt.tight_layout()
plt.show()
```

## 4. Key Findings and Conclusion

### Key Findings

- **Pricing:** Listing prices vary across the dataset, although the average prices across neighbourhood groups are relatively close.
- **Room types:** Entire homes/apartments and private rooms account for most listings.
- **Reviews:** Hotel rooms have the highest average number of reviews by room type, but their sample size is much smaller.
- **Availability:** Average availability differs by room type. Availability represents days listed as available, not the number of days booked.

### Limitations

- **Missing data:** Some listings contain missing values, including 247 listings with missing prices. These were excluded from the price-distribution analysis.
- **Data quality:** The original dataset contains inconsistent and suspicious values. Cleaning rules were applied, but some source-data limitations may remain.
- **Uneven sample sizes:** Hotel rooms are much less common than other room types, so their average review count should be interpreted cautiously.
- **Association versus causation:** Differences in price, reviews, and availability describe patterns in the dataset; they do not establish cause-and-effect relationships.
- **Dataset scope:** The analysis describes the available dataset and should not automatically be generalized to all Airbnb listings or other time periods.
  
### Conclusion

This project analyzed Airbnb listings using Python, Pandas, NumPy, and Matplotlib. The analysis explored listing prices, neighbourhood groups, room types, reviews, and availability. Data cleaning and exploratory data analysis helped identify patterns and data-quality issues. The findings should be interpreted with consideration for missing values, unusual records, and differences in sample sizes.
